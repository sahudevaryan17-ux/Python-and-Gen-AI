# How lists are stored in memory:-
1. Contiguous References – Stores object references in contiguous memory, not actual objects.
2. References – Each element is a reference to an object.
3. Dynamic – Automatically resizes when needed.
4. Mixed Types – Can store int, str, float, bool, etc., together.
5. Fast Access – Indexing is O(1).

```python
l = [1,2,3]
# id() id used to print address
print(id(l))
#list me vo element kaha stored hai
print(id(l[0]))
print(id(l[1]))
print(id(l[2]))
#khud element kaha stored hai
print(id(1))
print(id(2))
print(id(3))
```

```python
#order matters.
l1 = [1,2,3]
l2 = [3,2,1]
l1 == l2
```

# List Comprehension

1. provides concise way of creating list
2. more time efficient
3. space efficient
4. req few lines of code
5. jo kaam hum loop laga ke karenge usko hum fomula ki taraha kar lenge.

L = [expresssion (for) item (in) iterator (if) condition == true]


```python
# 1 to 10 numbers in list
L = [ i for i in range(1,100)]
print(L)
```

```python
# scaler mult 
v = [4,3,4,6,5]
s = -9
l = [s*i for i in v]
# item jo vector v me hai usko mult kardo s se 
print(l)
```

```python
l =[1,2,3,4,5]
L = [i*i for i in l]
print (L)
```

```python
# print all the numbers div by 5 in 1 to 50
[i for i in range(1,51) if i%5 == 0]
```

```python
#find lang which starts with p letter
langs = ['java','php','python','javascript']
[i for i in langs if i.startswith('p') ]
```

```python
l1 = [1,2,3,4]
l2 = [6,7,8,9]
[i*j for i in l1 for j in l2]
```
